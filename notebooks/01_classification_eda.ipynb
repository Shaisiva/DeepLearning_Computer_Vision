# Classification dataset — quick EDA

Copy `classification_dataset` to `data/classification_dataset/` with `TRAIN`, `VALID`, `TEST` and `bird` / `drone` subfolders.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if (Path.cwd().name == "notebooks") else Path.cwd()
sys.path.insert(0, str(ROOT))

from src import config

root = config.CLASSIFICATION_ROOT
print("Root:", root)

for split in ("TRAIN", "VALID", "TEST"):
    d = root / split
    if not d.is_dir():
        print(split, "MISSING")
        continue
    counts = {}
    for cls in sorted(p.name for p in d.iterdir() if p.is_dir()):
        n = sum(1 for p in (d / cls).rglob("*") if p.suffix.lower() in {".jpg", ".jpeg", ".png"})
        counts[cls] = n
    print(split, counts, "total", sum(counts.values()))
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import random

bird_dir = root / "TRAIN" / "bird"
drone_dir = root / "TRAIN" / "drone"
if bird_dir.is_dir() and drone_dir.is_dir():
    b_imgs = list(bird_dir.glob("*.jpg"))[:200]
    d_imgs = list(drone_dir.glob("*.jpg"))[:200]
    fig, axes = plt.subplots(2, 4, figsize=(10, 5))
    for row, (pool, title) in enumerate([(b_imgs, "bird"), (d_imgs, "drone")]):
        for col in range(4):
            ax = axes[row, col]
            p = random.choice(pool)
            ax.imshow(Image.open(p).convert("RGB"))
            ax.axis("off")
    plt.suptitle("Random training samples")
    plt.tight_layout()
    plt.show()
else:
    print("Add training images first.")
```
